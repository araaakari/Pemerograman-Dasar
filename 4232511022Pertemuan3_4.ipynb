```python
#soal 1 : input bertipe float
total = float(input("Berapa porsi makan yang Anda beli? : "))
Nama = str(input("Siapa Nama Pelanggan ? : "))

print("Total porsi :", total, type(total))
print("Nama        :", Nama, type(Nama))


#soal 2 : output frngsn karakter format
nama_makanan = "Mi ayam bakso"
jumlah_porsi = 3
harga_per_porsi = 15000.0
kode_hex = 255
kode_oktal = 64

print("Menu          : %s" % nama_makanan)
print("Jumlah porsi  : %d" % jumlah_porsi)
print("Harga/porsi   : %.2f" % harga_per_porsi)
print("Kode hexa     : %x" % kode_hex)
print("Kode oktal    : %o" % kode_oktal)


#soal 3 : diskon belanja
nama_barang = input("Nama barang   : ")
belanja = float(input("Total belanja : "))

diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon

print("Barang       :", nama_barang)
print("Diskon       : %.2f" % diskon)
print("Total harga  : %.2f" % total_harga)


#soal 4 : nilai akhir mahasiswa
tugas     = float(input("Nilai tugas     : "))
kuis      = float(input("Nilai kuis      : "))
praktikum = float(input("Nilai praktikum : "))
uts       = float(input("Nilai UTS       : "))
uas       = float(input("Nilai UAS       : "))
standar_kelulusan = float(input("Standar kelulusan: "))

nilai_mahasiswa = (0.2 * tugas) + (0.1 * kuis) + (0.2 * praktikum) + (0.2 * uts) + (0.3 * uas)

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa     : {nilai_mahasiswa}")
print(f"Standar Kelulusan   : {standar_kelulusan}")
print(f"Apakah Lulus?       : {apakah_lulus}")
print(f"Perlu Remedial?     : {perlu_remedial}")



#soal 5 : konversi total detik
try:
    total_detik = int(input("Masukkan total detik: "))
    if total_detik < 0:
        print("Total detik tidak boleh negatif")
    else:
        jam = total_detik // 3600
        sisa_detik = total_detik % 3600
        menit = sisa_detik // 60      # slide: // 6 (keliru)
        detik = sisa_detik % 60
        print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
except ValueError:
    print("Input harus berupa bilangan bulat")



#Latihan
#soal 1 : menentukan tahun kabisat
tahun = int(input("Masukkan tahun: "))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
print(f"Apakah {tahun} tahun kabisat? {kabisat}")


#soal 2 : kalkulator bmi
nama = input("Nama                : ")
berat = float(input("Berat badan (kg)    : "))
tinggi_cm = float(input("Tinggi badan (cm)   : "))

tinggi_m = tinggi_cm / 100
bmi = berat / (tinggi_m ** 2)
ideal = bmi >= 18.5 and bmi < 25

print("Halo %s" % nama)
print("BMI Anda : %.2f" % bmi)
print("Berat badan ideal?", ideal)

#soal di github
#soal 1 : belanja buku
jumlah = int(input("Jumlah buku    : "))
harga = int(input("Harga per buku : "))
uang = int(input("Uang Budi      : "))
orang = int(input("Jumlah orang   : "))
persen_diskon = int(input("Diskon (%)     : "))

total = jumlah * harga
diskon = total * persen_diskon // 100
bayar = total - diskon
kembalian = uang - bayar
per_orang = bayar // orang
sisa = bayar % orang
cukup = uang >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", cukup)



#soal 2 : PLTS atap rumah
panel = int(input("Jumlah panel       : "))
daya = float(input("Daya per panel (Wp): "))
jam = float(input("Jam penyinaran/hari: "))
efisiensi = float(input("Efisiensi (0-1)    : "))
kebutuhan = float(input("Kebutuhan (kWh/hari): "))
tarif = float(input("Tarif (Rp/kWh)     : "))

harian = panel * daya / 1000 * jam * efisiensi
bulanan = harian * 30
nilai = bulanan * tarif
kurang = kebutuhan - harian
cukup = harian >= kebutuhan

print("Energi harian (kWh) :", round(harian, 1))
print("Energi bulanan (kWh):", round(bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai, 1))
print("Kekurangan (kWh)    :", round(kurang, 1))
print("Mencukupi?          :", cukup)



#soal 3 : generator diesel
daya = float(input("Daya generator (kW)    : "))
jam = float(input("Jam operasi/hari      : "))
konsumsi = float(input("Konsumsi (liter/kWh)  : "))
harga = float(input("Harga solar (Rp/liter): "))
tangki = float(input("Kapasitas tangki (L)  : "))
arus = float(input("Arus kabel (A)        : "))
hambatan = float(input("Hambatan kabel (ohm)  : "))

energi = int(daya * jam)
solar = energi * konsumsi
biaya = solar * harga
hari = tangki // solar
sisa = tangki % solar
rugi = arus ** 2 * hambatan
tinggi = rugi > 150 and arus >= 20

print("Energi (kWh)  :", energi)
print("Solar (liter) :", solar)
print("Biaya (Rp)    :", biaya)
print("Tangki cukup  :", hari, "hari")
print("Sisa solar (L):", sisa)
print("Rugi kabel (W):", rugi)
print("Rugi tinggi?  :", tinggi)
```
